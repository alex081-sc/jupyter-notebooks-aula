### Questão 3: Distância entre dois pontos (1,0 ponto)

Usando a instrução `import` e o módulo `math`, escreva um programa que:

1. leia as coordenadas de dois pontos do plano: `x1`, `y1`, `x2` e `y2` (todos `float`);
2. calcule `dx` e `dy` como as diferenças **em valor absoluto** entre as coordenadas;
3. calcule a distância entre os pontos com a fórmula abaixo:

$$distancia = \sqrt{dx^2 + dy^2}$$

4. calcule a área do círculo que tem essa distância como raio;
5. exiba `dx`, `dy`, a distância e a área, todos arredondados para 2 casas decimais com `round`.

Exemplo:

```
x1: 1
y1: 2
x2: 4
y2: 6

dx = 3.0
dy = 4.0
Distância entre os pontos: 5.0
Área do círculo de raio 5.0: 78.54
```

````python
# Inclua aqui a resposta
import math
x1 = float(input("x1: "))
y1 = float(input("y1: "))
x2 = float(input("x2: "))
y2 = float(input("y2: "))
dx = abs(x1 - x2)
dy = abs(y1 - y2)
distancia = math.sqrt((dx ** 2)+ (dy ** 2))
area_circulo = math.pi*(distancia ** 2)

print(f"dx = {round(dx, 2)}")
print(f"dy = {round(dy, 2)}")
print(f"Distancia entre os pontos: {round(distancia, 2)}")
print(f"Area do circulo de raio {round(distancia, 2)}: {round(area_circulo, 2)}")
````

```python

```

Sua tarefa:

**(a)** **Encapsule** essa lógica em uma função chamada `exibir_media(nome, nota1, nota2, nota3)`, que calcula a
média das três notas e exibe a mensagem no mesmo formato. A função deve ter uma **docstring** descrevendo o que
ela faz e o que significa cada parâmetro.

**(b)** Chame a função três vezes para reproduzir exatamente a mesma saída do código original.

**(c)** Na célula de texto indicada, explique com suas palavras: o que é **encapsulamento**, o que é

Saída esperada:

```

```


````python
# Inclua aqui a resposta
def exibir_media(nome, nota1, nota2, nota3):
  """ Recebe nome e 3 notas.
  Calcula a soma das 3 notas e divide por 3 para obter a media.
  Exibe na tela a media com o nome do aluno.
  Paramentros:
    Nome: nome do aluno
    nota1, nota2 e nota3: Primeira, segunda e terceira nota do aluno, Respectivamente.
  """
  media = (nota1 + nota2 + nota3) / 3
  print(f"Media de {nome}: {round(media, 2)}")

exibir_media("Ana", 7.0, 5.5, 9.0)
exibir_media("Bruno", 4.0, 6.5, 5.0)
exibir_media("Carla", 9.5, 8.0, 10.0)
````

**(c) Sua explicação sobre encapsulamento e refatoração:**

_(  **Encapsulamento** é pegar uma parte de um código e juntar em um bloco de código. **Refatorar** é pegar um código e reorganizar ele, deixando mais limpo e facil de entender. A versão com função é melhor, porque evita a repetição de código, deixa o código mais legivel e organizado e facilita a manutenção, porque se acontecer um erro é só arrumar a função.)_

A função deve:

- imprimir o `texto` entre dois `caractere` (um no início e outro no fim da linha);

````python
# Inclua aqui a resposta
def moldura(texto, caractere, largura):
  """Recebe um texto, um carctere e um valor de largura.
  Exibe uma linha com o carctere, a linha tem o tamanho da largura. Ela é exibida no começo e no fim da função
  Exibe o texto entre o caractere.
  Parametros:
    texto: texto que vai ser exibido no meio da moldura
    carctere: carctere que vai ser usado para criar a linha que vai servir como moldura
    largura: valor que vai definir a quantidade de vezes que o carctere vai se repetir na linha
  """
  print(caractere*largura)
  print(f"{caractere} {texto} {caractere}")
  print(caractere*largura)

moldura("Trabalho 1", "=", 20)
moldura("Trabalho 3", "*", 30)
moldura("Trabalho 2", "-", 25)
````

### Questão 6: Desenhando com `jupyturtle` (1,0 ponto)

Não esqueça de baixar o módulo `jupyturtle` antes de resolver a questão.


**(a)** Escreva uma função chamada `poligono(lados, tamanho)` que desenhe um polígono regular com o número de lados e o comprimento de lado informados. A função deve ter docstring.

**(b)** Escreva uma função chamada `flor(petalas, lados, tamanho)` que **reutilize** `poligono` para desenhar
uma flor.

**(c)** Faça três desenhos: um triângulo, um hexágono e uma flor.

Exemplo:

```
make_turtle()
flor(8, 4, 30)
```

````python
from os.path import basename, exists

def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve

        local, _ = urlretrieve(url, filename)
        print("Download realizado: " + str(local))
    return filename

download('https://github.com/ramalho/jupyturtle/releases/download/2024-03/jupyturtle.py');
````

````python
# (a) e (b) defina aqui as funções poligono e flor
from jupyturtle import make_turtle, forward, left
def poligono(lados, tamanho):
  """Desenha um poligono com lados e tamanho especificos
  parametros:
    lados: quantos lados o poligono terá.
    tamanho: tamanho de cada lado
  """
  angulo = 360 / lados
  for i in range(lados):
    forward(tamanho)
    left(angulo)

def flor(petalas, lados, tamanho):
  """Recebe a quantidade de petalas, o lado e o tamanho.
  Para cada petala, a função vai desenhar um poligono com a quantidade de lados
  e tamanho passados por parametro
  paramentors:
    petalas: quantidade de poligonos que formarao a flor.
    lados: quantos lados cada poligono tera
    tamanho: tamanho de cada lado do poligono
  """
  angulo = 360 / petalas
  for i in range(petalas):
    poligono(lados, tamanho)
    left(angulo)
````

````python
# (c) triângulo
make_turtle()
poligono(3,30)
````

````python
# (c) hexágono
make_turtle()
poligono(6,30)
````

````python
# (c) flor
make_turtle()
flor(4, 6, 20)
````

### Questão 7: Funções booleanas e operadores lógicos (1,0 ponto)

**(a)** Escreva uma função chamada `eh_bissexto(ano)` que **retorne** `True` se o ano for bissexto e `False` caso contrário. Use os operadores `%`, `and`, `or` e `not` em uma **única expressão booleana** no `return`
(não use `if`).

**(b)** Escreva uma função chamada `eh_triangulo(a, b, c)` que retorne `True` se três medidas podem formar um
triângulo. A condição é que cada lado seja **positivo** e que cada lado seja **menor** que a soma dos outros dois.

**(c)** Usando `for i in range(...)`, exiba para os anos de **1996 a 2005** se cada um é bissexto ou não.

**(d)** Teste `eh_triangulo` com pelo menos três combinações diferentes, exibindo o resultado de cada teste.

Exemplo:

```
1996 é bissexto? True
1997 é bissexto? False
...
2000 é bissexto? True
...
eh_triangulo(3, 4, 5) -> True
eh_triangulo(1, 2, 10) -> False
eh_triangulo(-1, 4, 5) -> False
```

````python
# Inclua aqui a resposta
def eh_bissexto(ano):
  """Recebe o ano e faz um calculo para descobrir se o ano é bissexto ou não,
  retorna False caso nao seja bissexto, e True se for bissexto.
  """
  return ((ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0))

def eh_triangulo(a, b, c):
  """Recebe a, b e c, e verifica se esses valores podem formar um triangulo.
  retorna True se pudere formar um triangulo, e False caso seja o contrario
  """
  if(a <= 0 or b <= 0 or c <= 0):
    return False
  elif((a+b <= c) or (b+c <= a) or (c+a <= b)):
    return False
  return True

for i in range(1996, 2006):
  print(f"{i} é bissexto? {eh_bissexto(i)}")

print("\n")
print(eh_triangulo(3,4,5))
print(eh_triangulo(1,2,10))
print(eh_triangulo(-1,4,5))
````

### Questão 8: Condicionais encadeadas e aninhadas (1,0 ponto)

Escreva uma função chamada `classificar_imc(peso, altura)` que:

1. **valide os dados**: se `peso` **ou** `altura` for menor ou igual a zero, a função deve
   retornar `None`, sem calcular nada;
2. caso os dados sejam válidos, calcule o IMC com a fórmula `imc = peso / altura ** 2`;
3. use uma **condicional encadeada** (`if`/`elif`/`else`) para **retornar** a classificação correspondente:

| IMC | Classificação |
|---|---|
| menor que 18,5 | `"Abaixo do peso"` |
| de 18,5 a 24,9 | `"Peso normal"` |
| de 25,0 a 29,9 | `"Sobrepeso"` |
| 30,0 ou mais | `"Obesidade"` |

Depois, escreva um programa que leia o peso e a altura do usuário, chame a função e:

- se o retorno for `None`, exiba a mensagem `"Dados inválidos: peso e altura devem ser positivos."`;
- caso contrário, exiba o IMC arredondado com 2 casas decimais e a classificação.

A função deve ter **docstring**.

Exemplo:

```
Peso (kg): 72.5
Altura (m): 1.75
IMC: 23.67 -> Peso normal
```

```
Peso (kg): 70
Altura (m): 0
Dados inválidos: peso e altura devem ser positivos.
```

````python
# Inclua aqui a resposta

def classificar_imc(peso, altura):
  """Recebe peso e altura. Usa a formula imc = peso / altura ** 2 para descobrir
  o imc.
  parametros:
    peso: peso da pessoa em kgs.
    altura: altura da pessoa em metros. ex(1.8 == 180) metros/centimetros
  """
  if(peso <= 0 or altura <= 0):
    return None
  else:
    imc = peso / altura ** 2

  if(imc < 18.5):
    return   f"{round(imc,2)} -> Abaixo do Peso"
  elif(imc < 25):
    return f"{round(imc,2)} -> Peso normal"
  elif(imc < 30):
    return f"{round(imc,2)} -> Sobrepeso"
  else:
    return f"{round(imc,2)} -> Obesidade"

peso = float(input("Peso(kg): "))
altura = float(input("Altura(m): "))
classificacao = classificar_imc(peso, altura)
if(classificacao == None):
  print("Dados inválidos: peso e altura devem ser positivos.")
else:
  print(classificacao)
````

````python
# Versão 1 - somente as funções, testadas com valores fixos

def media_notas(nota1, nota2, nota3):
  """ Recebe as 3 notas do aluno <float> e devolve a media dessas notas <float>
  """
  return round((nota1+nota2+nota3)/3, 2)

def situacao(media):
  """Recebe a media <float> e retorna um texto, podendo ser: Aprovado,
  Reprovado ou Recuperacao.
  """
  if(media >= 7):
    return "Aprovado"
  elif(media >= 4):
    return "Recuperacao"
  else:
    return "Reprovado"

print(media_notas(7,5.5,9))
print(situacao(7))
print(situacao(6))
print(situacao(3))
````

````python
# Versão 2 - laço de leitura e linhas do boletim
alunos = []
qtd_alunos = int(input("Quantidade de alunos: "))

for i in range(qtd_alunos):
  nome = input("Nome do aluno: ")
  nota1 = float(input("Primeira nota: "))
  nota2 = float(input("Segunda nota: "))
  nota3 = float(input("Terceira nota: "))
  media = media_notas(nota1, nota2, nota3)
  situaca = situacao(media)
  alunos.append({"nome": nome, "media": media, "situaca": situaca})

print("\n==== BOLETIM ====")
for aluno in alunos:
  print(f"{aluno['nome']}: {aluno['media']} - {aluno['situaca']}")
````

````python
# Versão 3 - versão final, com as estatísticas da turma
alunos = []
qtd_alunos = int(input("Quantidade de alunos: "))

for i in range(qtd_alunos):
  nome = input("Nome do aluno: ")
  nota1 = float(input("Primeira nota: "))
  nota2 = float(input("Segunda nota: "))
  nota3 = float(input("Terceira nota: "))
  media = media_notas(nota1, nota2, nota3)
  situaca = situacao(media)
  alunos.append({"nome": nome, "media": media, "situaca": situaca})

print("\n==== BOLETIM ====")
for aluno in alunos:
  print(f"{aluno['nome']}: {aluno['media']} - {aluno['situaca']}")

media_geral = 0
aprovados = 0
maior_media = 0
aluno_maior = ""
for aluno in alunos:
  media_geral+= aluno['media']
  if(aluno['situaca'] == "Aprovado"):
    aprovados+=1
  if(maior_media < aluno['media']):
    maior_media = aluno['media']
    aluno_maior = aluno['nome']
media_geral/= qtd_alunos

print("\n======================")
print(f"Média geral da turma: {round(media_geral, 2)}")
print(f"Aprovados: {aprovados} de {qtd_alunos} ({round((aprovados/qtd_alunos)*100, 2)}%)")
print(f"Melhor média: {aluno_maior} ({maior_media})")
````
